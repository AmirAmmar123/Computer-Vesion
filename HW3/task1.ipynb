
## Introduction
This project explores the characteristics and effects of different aperture shapes on the images produced by a pinhole camera.

## Ideal Pinhole Diameter
The ideal pinhole diameter is defined by the following equation:
d = 2 * sqrt( f * λ )

where `λ` is the wavelength and `f` is the effective focal length.

## Aperture Shapes and Their Effects

### a) Small Circular Aperture
- **Image Characteristics**: The image will be sharp but relatively dim.
- **Explanation**: A small circular aperture allows only a small amount of light to pass through, which results in a sharp but dim image. The small aperture ensures that light rays from a single point on the scene converge at a single point on the image plane, minimizing blur.

### b) Large Circular Aperture
- **Image Characteristics**: The image will be brighter but blurrier.
- **Explanation**: A large circular aperture allows more light to pass through, which results in a brighter image. However, the increased aperture size means that light rays from a single point on the scene spread out over a larger area on the image plane, causing the image to be less sharp and blurrier.

### c) Vertical Slit Aperture
- **Image Characteristics**: The image will be sharp in the horizontal direction but blurred in the vertical direction.
- **Explanation**: A vertical slit aperture restricts the light rays horizontally, resulting in a sharp image in the horizontal direction. However, it allows light rays to spread more in the vertical direction, causing vertical blur.

### d) Horizontal Slit Aperture
- **Image Characteristics**: The image will be sharp in the vertical direction but blurred in the horizontal direction.
- **Explanation**: A horizontal slit aperture restricts the light rays vertically, resulting in a sharp image in the vertical direction. However, it allows light rays to spread more in the horizontal direction, causing horizontal blur.

## Problems with Very Small or Very Large Apertures

### Very Small Aperture
- **Problem**: Diffraction.
- **Explanation**: When the aperture is very small, diffraction effects become significant. Diffraction causes light to bend around the edges of the aperture, which can blur the image and reduce its sharpness. Additionally, the image will be very dim because only a small amount of light can pass through.

### Very Large Aperture
- **Problem**: Increased blur due to large aperture.
- **Explanation**: A very large aperture allows more light rays from different points in the scene to converge at different points on the image plane, leading to a significant increase in blur. The image will be brighter, but the lack of convergence will make it much less sharp and can cause overlapping of light rays from different points in the scene, resulting in a loss of detail.
